```python
spam_words = [
    "lottery",
    "claim",
    "click",
    "winner",
    "prize",
    "free",
    "lobbery",
    "money",
]

email = input("Enter your email message:\n")
email = email.lower()

spam = False

for word in spam_words:
    if word in email:
        spam = True
        break

if spam:
    print("\nResult: Spam Mail")
else:
    print("\nResult: Not Spam Mail")
```

```python
print("Create a 🤖 Chatbot:")

while True:
    user = input("You: ").lower()

    if user == "hi" or user == "hello":
        print("🤖 Chatbot: Hello! How are you?")

    elif "how are you" in user:
        print("🤖 Chatbot: I'm fine, thank you!")

    elif "what is your name" in user:
        print("🤖 Chatbot: My name is Chatbot.")

    elif user == "bye":
        print("🤖 Chatbot: Bye!")
        break

    else:
        print("🤖 Chatbot: I don't understand.")
```
